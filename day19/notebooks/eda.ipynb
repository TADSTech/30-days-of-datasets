```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../data/StudentPerformanceFactors.csv')
```

```python
df.head()

df.info()

df.describe()
```

## Data Overview

The dataset contains 6,607 student records with 20 features including demographic, academic, and environmental factors. The average exam score is 67.24 with a standard deviation of 3.89. Scores range from 55 to 101. There are some missing values in Teacher_Quality, Parental_Education_Level, and Distance_from_Home.

```python
fig = px.histogram(df, x='Exam_Score', nbins=20, title='Distribution of Exam Scores')
fig.show()
```

## Distribution of Exam Scores

The exam scores are approximately normally distributed, with most students scoring between 65 and 69. The median score is 67.

```python
fig = px.bar(df['Gender'].value_counts(), title='Gender Distribution')
fig.show()
```

## Gender Distribution

The dataset includes 3,814 male and 2,793 female students, with a slight majority of males.

```python
fig = px.box(df, x='Gender', y='Exam_Score', title='Exam Scores by Gender')
fig.show()
```

## Exam Scores by Gender

Male and female students show similar distributions in exam scores, with comparable medians and ranges.

```python
fig = px.box(df, x='Parental_Education_Level', y='Exam_Score', title='Exam Scores by Parental Education Level')
fig.show()
```

## Exam Scores by Parental Education Level

Students whose parents have postgraduate education show slightly higher median exam scores compared to those with high school or college-educated parents.

```python
numeric_cols = df.select_dtypes(include=[np.number]).columns
corr_matrix = df[numeric_cols].corr()
fig = px.imshow(corr_matrix, text_auto=True, title='Correlation Matrix')
fig.show()
```

## Correlation Analysis

Attendance has the strongest positive correlation with exam scores (0.58), followed by hours studied (0.45). Previous scores and tutoring sessions show moderate positive correlations, while sleep hours and physical activity have weak correlations.
